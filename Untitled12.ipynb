```python
import pandas as pd
import ipywidgets as widgets
from IPython.display import display

# Criando um DataFrame para veículos
df_veiculos = pd.DataFrame(columns=['ID', 'Modelo', 'Ano', 'Placa', 'Manutencao', 'Data'])

import pandas as pd
import ipywidgets as widgets
from IPython.display import display

# Criando um DataFrame para veículos
df_veiculos = pd.DataFrame(columns=['ID', 'Modelo', 'Ano', 'Placa', 'Manutencao', 'Data'])

# Função para adicionar veículos
def adicionar_veiculo(modelo, ano, placa):
    global df_veiculos
    novo_id = len(df_veiculos) + 1
    df_veiculos.loc[len(df_veiculos)] = [novo_id, modelo, ano, placa, '', '']
    print(f"Veículo {modelo} adicionado com sucesso!")
    exibir_dados()

# Função para registrar manutenções
def registrar_manutencao(id_veiculo, manutencao, data):
    global df_veiculos
    if id_veiculo in df_veiculos['ID'].values:
        df_veiculos.loc[df_veiculos['ID'] == id_veiculo, 'Manutencao'] = manutencao
        df_veiculos.loc[df_veiculos['ID'] == id_veiculo, 'Data'] = data
        print(f"Manutenção registrada para o veículo ID {id_veiculo}.")
    else:
        print(f"Veículo com ID {id_veiculo} não encontrado.")
    exibir_dados()

# Função para criar o formulário de adição de veículo
def formulario_adicionar_veiculo():
    modelo = widgets.Text(description='Modelo:')
    ano = widgets.IntText(description='Ano:')
    placa = widgets.Text(description='Placa:')
    button = widgets.Button(description='Adicionar Veículo')

    def on_button_click(b):
        if modelo.value and ano.value > 1885 and placa.value:  # O primeiro carro foi em 1886
            adicionar_veiculo(modelo.value, ano.value, placa.value)
            modelo.value = ''
            ano.value = 0  # Reseta para um número válido
            placa.value = ''
        else:
            print("Por favor, preencha todos os campos corretamente. O ano deve ser um número maior que 1885.")

    button.on_click(on_button_click)
    display(modelo, ano, placa, button)

# Função para criar o formulário de registro de manutenção
def formulario_registrar_manutencao():
    id_veiculo = widgets.IntText(description='ID do Veículo:')
    manutencao = widgets.Text(description='Manutenção:')
    data = widgets.Text(description='Data (YYYY-MM-DD):')
    button = widgets.Button(description='Registrar Manutenção')

    def on_button_click(b):
        if id_veiculo.value and manutencao.value and data.value:
            registrar_manutencao(id_veiculo.value, manutencao.value, data.value)
            id_veiculo.value = 0  # Reseta para um número válido
            manutencao.value = ''
            data.value = ''
        else:
            print("Por favor, preencha todos os campos.")

    button.on_click(on_button_click)
    display(id_veiculo, manutencao, data, button)

# Função para exibir os dados
def exibir_dados():
    print("\nDados dos veículos:")
    print(df_veiculos)

# Exibindo os formulários
print("Formulário para adicionar veículos:")
formulario_adicionar_veiculo()

print("\nFormulário para registrar manutenções:")
formulario_registrar_manutencao()
```
